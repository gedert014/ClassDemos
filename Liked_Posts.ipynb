```python
tidy_likes = []

for post in liked_posts:
    timestamp = post["timestamp"]
    caption = ""
    name = ""
    username = ""

    for item in post["label_values"]:

        if item.get("label") == "Caption":
            caption = item["value"]

        if item.get("title") == "Owner":
            owner = item["dict"][0]["dict"]

            for detail in owner:
                if detail["label"] == "Name":
                    name = detail["value"]

                if detail["label"] == "Username":
                    username = detail["value"]

    tidy_likes.append({
        "timestamp": timestamp,
        "caption": caption,
        "name": name,
        "username": username
    })
```

```python
likes_df = pd.DataFrame(tidy_likes)

likes_df["timestamp"] = pd.to_datetime(
likes_df["timestamp"],
unit="s"
)

likes_df.head()
```

```python
likes_df.groupby("username").count()
```

```python
timestamps = [
    post["timestamp"]
    for post in liked_posts
]

captions = [
    post["label_values"][1]["value"]
    for post in liked_posts
]

usernames = [
    post["label_values"][4]["dict"][0]["dict"][2]["value"]
    for post in liked_posts
]

likes_df = pd.DataFrame({
    "timestamp": timestamps,
    "caption": captions,
    "username": usernames
})

likes_df["timestamp"] = pd.to_datetime(
    likes_df["timestamp"],
    unit="s"
)

likes_df.head()
```

```python
timestamps = []
captions = []
usernames = []

for post in liked_posts:
    
    timestamp = post["timestamp"]
    caption = ""
    username = ""
    
    for item in post["label_values"]:
        
        if item.get("label") == "Caption":
            caption = item.get("value", "")
            
        if item.get("title") == "Owner":
            for owner in item.get("dict", []):
                for detail in owner.get("dict", []):
                    if detail.get("label") == "Username":
                        username = detail.get("value", "")
    
    timestamps.append(timestamp)
    captions.append(caption)
    usernames.append(username)


likes_df = pd.DataFrame({
    "timestamp": timestamps,
    "caption": captions,
    "username": usernames
})


likes_df["timestamp"] = pd.to_datetime(
    likes_df["timestamp"],
    unit="s"
)


likes_df.head()
```

```python
likes_by_account = (
    likes_df.groupby("username")["timestamp"]
    .count()
    .sort_values(ascending=False)
)

likes_by_account.head(10)
```
